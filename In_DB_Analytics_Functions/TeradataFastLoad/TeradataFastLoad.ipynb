<h4>1. Import the required libraries</h4>
<p>Create a settings.py file that contains your Vantage DB credentials</p>

```python
# Import the required libraries
from teradataml import create_context, DataFrame
import subprocess
import os

import settings
```

<h4>2. Change the directory to where the fastload.exe file is located</h4>
<b>Change the fastload_path value to the location of fastload.exe on your machine</b>

```python
# Change the directory to where fastload.exe is located
fastload_path = r"C:\Program Files\Teradata\Client\20.00\bin"
os.chdir(fastload_path)
```

<h4>3. Run the dataload script in Fastload</h4>
<b>Change the fastload_script value to the location of your fastload script</b>

```python
# Save the fastload script path to a variable
fastload_script = r"C:/path/to/fastload/script"

# Execute the FastLoad script
process = subprocess.Popen(
    ["fastload"],
    stdin=open(fastload_script, 'r'),
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    text=True
)
```

<h4>4. View the results</h4>

```python
# Capture output and errors
stdout, stderr = process.communicate()

# Display the results
print("STDOUT:")
print(stdout)
print("STDERR:")
print(stderr)
```

<h5>Connect to the database</h5>

```python
# Connect to the database
create_context(host=settings.system_dsn, username=settings.system_user, password=settings.system_pw, logmech='LDAP')
```

<h4>5. Check the table to ensure the data was inserted correctly</h4>
