<b style = 'font-size:30px;font-family:Arial;color:#00233C'>2. Unstructured.io Partitioning into JSON </b>

### 🔮 Using Unstructured.io Platform

We can leverage the Unstructured Partition Endpoint to structure our file into a standard canonical JSON format

1. **Use the partition function**
   - Insert your personal API key
   - Specify our intput file and the name of our output file
   - Then, we'll issue a partitioning request from the API, sending it our source document.
      - We'll need to choose a partition strategy.

<div style="font-size:15px;font-family:Arial;color:#00233C;background:#fff8e6;padding:12px;border-radius:5px;border:1px solid #ffd700">
<b>Note:</b> To run the notebook successfully, you'll need to grab your API Key by logging into Unstructured (https://platform.unstructured.io/app/start) then navigate to API Keys in the left-side navigation. Copy your API Key and paste it into the .env file where it currently reads UNSTRUCTURED_API_KEY=your-api-key-here.
</div>

```python
# Grab your API Key from the Unstructured Platform
load_dotenv('.env')
client = unstructured_client.UnstructuredClient(
    api_key_auth=os.getenv("UNSTRUCTURED_API_KEY") 
)

input_file = r"DensePassageRetrieval.pdf"
output_file = r"UnstructuredJSON.json"


req = operations.PartitionRequest(
    partition_parameters=shared.PartitionParameters(
        files=shared.Files(
            content=open(input_file, "rb"),
            file_name=input_file,
        ),
        strategy=shared.Strategy.AUTO,   # Select your strategy.
        vlm_model="gpt-4o",             # If the Auto router decides to use VLM, this is the model to use.
        vlm_model_provider="openai",    # If the Auto router decides to use VLM, this is the provider to use.
        languages=['eng'],             
        split_pdf_page=True,            # If True, splits the PDF file into smaller chunks of pages.
        split_pdf_allow_failed=True,    # If True, the partitioning continues even if some pages fail.
        split_pdf_concurrency_level=15  # Set the number of concurrent request to the maximum value: 15.
    ),
)

try:
    res = client.general.partition(
        request=req
    )
    element_dicts = [element for element in res.elements]
    
    # Print the processed data's first element only.
    print(element_dicts[0])

    # Write the processed data to a local file.
    json_elements = json.dumps(element_dicts, indent=2)

    with open(output_file, "w") as file:
        file.write(json_elements)
except Exception as e:
    print(e)
```

<hr style="height:2px;border:none;background-color:#00233C;">
<b style = 'font-size:30px;font-family:Arial;color:#00233C'>3. Generate Embeddings </b>

### 🧩 Adding embeddings to the JSON file.  

1. **Load the JSON file and add embeddings to each element.**  
   - Process each element in the JSON file.
   - Save the updated JSON back into the original file.


```python
if __name__ == "__main__":
    embeddings = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2"
    )
    
    # Get the JSON file's path.
    if len(sys.argv) < 2:
        print("Error: Specify the path to the input JSON file.")
        print("For example, 'python embeddings.py myfile.json'")
        sys.exit(1)

    file_path = r"UnstructuredJSON.json"
    
    try:
        # Get the JSON file's contents.
        with open(file_path, 'r') as file:
            file_elements = json.load(file)

        for element in file_elements:
            # Get the element's "text" field.
            text = element["text"]
            # Generate the embeddings for that "text" field.
            query_result = embeddings.embed_query(text)
            # Add the embeddings to that element as an "embeddings" field.
            element["embeddings"] = query_result


        # Save the updated JSON back into the original file.
        with open(file_path, 'w') as file:
            json.dump(file_elements, file, indent=2)

        print(f"Done! Updated JSON saved to '{file_path}'.")

    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")
    except IOError:
        print(f"Error: Unable to access file '{file_path}'.")
```

<b style = 'font-size:30px;font-family:Arial;color:#00233C'>5. Import Embeddings to a Table </b>

### 🚚 Read in external embeddings and clean the input data.  

1. **Define function to clean the input data** 
   - This function removes any non-ASCII characters from the text, ensuring that only clean, compatible strings are inserted into the database.
2. **Create `manual_vector_table` to store text and ambeddings**  
   - This table is designed to store the filename, text chunks, the embedding vector, and a message field. 
   - The Embedding column is set up to hold the vector type data for each text chunk.
3. **Open the JSON file and insert the text and embeddings**
   - We open the processed JSON file, which contains a text field and an embeddings field. For each element, we:
      - Clean the text using the clean_data function.
      - Insert both the cleaned text and the embedding string into the manual_vector_table using an SQL INSERT statement.

```python
#Clean data function
import re
def clean_data(data):
    # Remove non-ASCII characters
    return re.sub(r'[^\x00-\x7F]+', '', data)
```

### ⛏️ Create the table to store the embeddings

```python
#Create a multiset table to store the vector data
create_normalized_table_query = '''
CREATE MULTISET TABLE manual_vector_table, FALLBACK,
    NO BEFORE JOURNAL,
    NO AFTER JOURNAL,
    CHECKSUM = DEFAULT,
    DEFAULT MERGEBLOCKRATIO,
    MAP = TD_MAP2
    (
        TD_ID INTEGER GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
        TD_FILENAME VARCHAR(1024) CHARACTER SET LATIN NOT CASESPECIFIC,
        CHUNKS VARCHAR(32000) CHARACTER SET LATIN NOT CASESPECIFIC,
        Embedding VECTOR,
        Message VARCHAR(32000) CHARACTER SET LATIN NOT CASESPECIFIC)
    PRIMARY INDEX (TD_ID);
'''

#Execute the query to create the table
execute_sql(create_normalized_table_query)
```

### ⛳ Insert embeddings into the table

```python
# Open the JSON file and insert the text and embeddings into 'manual_vector_table' table
with open('UnstructuredJSON.json', 'r', encoding='utf-8') as f:
    elements = json.load(f)

for element in elements:
    text = element.get('text', '')
    embeddings = element.get('embeddings', [])
    if text and embeddings:
        cleaned_text = clean_data(str(text)).replace("'", "''")
        # Convert embeddings list to comma-separated string
        embedding_str = ','.join(map(str, embeddings))
        insert_query = f"""
        INSERT INTO manual_vector_table (TD_FILENAME, CHUNKS, Embedding, Message)
        VALUES ('DensePassageRetrieval', '{cleaned_text}', '{embedding_str}', 'Inserted by {username}');
        """
        execute_sql(insert_query)
        print(f"Inserted element: {cleaned_text[:50]}... Embedding: {embedding_str[:50]}...")
```

### 🔎 View embeddings table

<b style = 'font-size:30px;font-family:Arial;color:#00233C'>6. Normalize the embeddings </b>

### 🎚️ Apply the vector normalization function from Teradata

- Create a table with data
- Store the normalized vector embeddings in the table

```python
#Normalize the embeddings
normalize_query = '''
CREATE TABLE manual_vector_table_normalized AS (
SELECT *
FROM TD_Vectornormalize(
    ON manual_vector_table AS InputTable
    USING
        IDColumns('TD_ID')
        TargetColumns('Embedding')
        Accumulate('TD_FILENAME', 'Message', 'CHUNKS')
        Approach('UNITVECTOR')
        EmbeddingSize(384)
) AS DT) WITH DATA;
'''

#Execute the query to normalize the embeddings
execute_sql(normalize_query)
```

### 🔎 Inspect the normalized embedding table

```python
#View the normalized data
view_normalized_query = '''
SELECT TOP 20 *
FROM manual_vector_table_normalized;
'''
#Execute the query to view the normalized data
#Test dataframe
test_df = DataFrame.from_query(view_normalized_query)
test_df.head(3)
```

<b style = 'font-size:30px;font-family:Arial;color:#00233C'>7. Semantic Similarity Search </b>

### ↔️ Vector Distance algorithm example

The vector distance algorithm measures similarity between vectors using mathematical operations like cosine similarity or Euclidean distance. It calculates how "close" two vectors are in the embedding space to find semantically related content.


1. **Embed the query**
    - Load the Hugging Face model (same as original embeddings) 
    - Generate query embedding 

```python
model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

# Generate query embedding 
query_text = "What is Open-domain Question Answering?"  
query_embedding = model.embed_query(query_text)

# Convert embedding to SQL-compatible string
query_embedding_str = clean_data(str(query_embedding).replace("'", "''").replace("[", "").replace("]", ""))
print(query_embedding_str)
```

2. **Create a table to store the query**
    - The TD_Vectordistance requires the input to be in a table formatted the same as the reference table. 

```python
create_query_embedding_table = '''
CREATE MULTISET TABLE query_table, FALLBACK,
    NO BEFORE JOURNAL,
    NO AFTER JOURNAL,
    CHECKSUM = DEFAULT,
    DEFAULT MERGEBLOCKRATIO,
    MAP = TD_MAP2
    (
        TD_ID INTEGER GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
        TD_FILENAME VARCHAR(1024) CHARACTER SET LATIN NOT CASESPECIFIC,
        CHUNKS VARCHAR(32000) CHARACTER SET LATIN NOT CASESPECIFIC,
        Embedding VECTOR,
        Message VARCHAR(32000) CHARACTER SET LATIN NOT CASESPECIFIC)
    PRIMARY INDEX (TD_ID);
'''

#Execute the query to create the table
execute_sql(create_query_embedding_table)
```

3. **Insert the query into the table**
    - Fill the table with the query embedding.

```python
insert_query = f'''
INSERT INTO query_table (TD_FILENAME, CHUNKS, Embedding, Message)
VALUES ('test.txt', 'This is a test message', '{query_embedding_str}', 'This is a test message');
'''
#Execute the query to insert the data
execute_sql(insert_query)
```

4. **Normalize the query embedding**
    - For better results in the semantic similarity search, we need to normalize the query embedding like we did with the PDF vector embeddings

```python
normalize_query_embedding = '''
CREATE TABLE query_table_normalized AS (
SELECT *
FROM TD_Vectornormalize(
    ON query_table AS InputTable
        USING
        IDColumns('TD_ID')
        TargetColumns('Embedding')
        Accumulate('TD_FILENAME', 'CHUNKS', 'Message')
        Approach('UNITVECTOR')
        EmbeddingSize(384)
) AS DT) WITH DATA;
'''

#Execute the query to normalize the query embedding
execute_sql(normalize_query_embedding)
```

5. **Run a Vector Distance on the query and the embeddings**
    - Select target and reference IDs, and distance type for the TD_Vectordistance function
    - Show the top 5 closest embeddings to the query

```python
#Run semantic similarity search using vector distance
VectorDistance_query_search = '''
SELECT target_id, reference_id, distancetype, cast(distance as decimal(36,15)) as distance FROM TD_VECTORDISTANCE (
ON query_table_normalized as TargetTable
ON manual_vector_table_normalized as ReferenceTable Dimension
USING
TargetIDColumn('TD_ID')
TargetFeatureColumns('Embedding')
RefIDColumn('TD_ID')
RefFeatureColumns('Embedding')
DistanceMeasure('euclidean')
topk(5)
EmbeddingSize(384)
) as dt;
'''

#execute_sql(VectorDistance_query)

test_df2 = DataFrame.from_query(VectorDistance_query_search)
test_df2.head(20)
```

6. **Show the text results of the vector distance search**
    - The IDs of the top 5 results are stored in top5
    - View the top5 results using `SELECT`

```python
#Get the top 5 results
test_df2_pd = test_df2.to_pandas()
top5 = ','.join(str(i) for i in test_df2_pd.nsmallest(5, 'distance')['reference_id'])

#Query the table to get the results
query_results = f'''
SELECT TD_ID, CHUNKS
FROM manual_vector_table_normalized
WHERE TD_ID IN ({top5})
'''
#Execute the query to get the results
getResults = DataFrame.from_query(query_results)
getResults.head(5)
```
