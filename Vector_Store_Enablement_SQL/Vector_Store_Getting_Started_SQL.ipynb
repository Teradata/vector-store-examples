<p style="font-size:24px;font-family:Arial;color:#00233C"><b>Introduction</b></p>

<p style="font-size:16px;font-family:Arial;color:#00233C">
Vector stores are specialized databases designed for efficient storage, indexing, and searching of high-dimensional vector embeddings. These embeddings, generated by AI models, enable similarity search and power applications in machine learning, NLP, recommendation systems, and image/video retrieval.
</p>

<p style="font-size:16px;font-family:Arial;color:#00233C">

<p style="font-size:16px;font-family:Arial;color:#00233C">

<p style="font-size:20px;font-family:Arial;color:#00233C"><b>Vector Search Algorithms</b></p>

<p style="font-size:16px;font-family:Arial;color:#00233C">
Teradata supports three main search algorithms:
</p>

<p style="font-size:18px;font-family:Arial;color:#00233C"><b>1. VectorDistance</b></p>
<ul style="font-size:16px;font-family:Arial;color:#00233C">
  <li><b>Flat index</b> (performs a full-table scan)</li>
  <li><b>Scalable and parallel search</b></li>
  <li><b>Exact search approach</b></li>
  <li>Suitable for <b>smaller datasets (<10K documents)</b></li>
</ul>

<p style="font-size:18px;font-family:Arial;color:#00233C"><b>2. Kmeans</b></p>
<ul style="font-size:16px;font-family:Arial;color:#00233C">
  <li><b>IVF (Inverted Vector File) index</b> (organizes data into clusters)</li>
  <li><b>Scalable and parallel training/search</b></li>
  <li><b>Approximate search approach</b></li>
  <li>Suitable for <b>larger datasets (>10K documents)</b></li>
</ul>

<p style="font-size:18px;font-family:Arial;color:#00233C"><b>3. HNSW</b></p>
<ul style="font-size:16px;font-family:Arial;color:#00233C">
  <li><b>HNSW (Hierarchical Navigable Small World) index</b></li>
  <li><b>Graph-based Approximate Nearest Neighbors (ANN) search</b></li>
  <li><b>Scalable and parallel training/search</b></li>
  <li><b>Approximate search approach</b></li>
  <li>Suitable for <b>larger datasets (>10K documents)</b></li>
</ul>

<b style = 'font-size:30px;font-family:Arial;color:#00233C'>2. Getting Started with Vector Store </b>

### 🛠 Steps to Create a Vector Table in SQL  

1. **Create a table to store data**  
   - Use `CREATE TABLE` with columns `id` and `text`


```
CREATE TABLE text_data (
    id INTEGER GENERATED ALWAYS AS IDENTITY NOT NULL,
    text VARCHAR(1000),
    PRIMARY KEY (id)
);
```

2. **Insert data into that table**  
   - Use `INSERT INTO` with example data to populate the table 
   - And then we can just view the data

```
INSERT INTO text_data (text) VALUES
('The quick brown fox jumps over the lazy dog.');
INSERT INTO text_data (text) VALUES
('A journey of a thousand miles begins with a single step.');
INSERT INTO text_data (text) VALUES
('To be or not to be, that is the question.');
INSERT INTO text_data (text) VALUES
('All that glitters is not gold.');
INSERT INTO text_data (text) VALUES
('The only thing we have to fear is fear itself.');
INSERT INTO text_data (text) VALUES
('In the end, we will remember not the words of our enemies, but the silence of our friends.');
```

```
-- View data in table
SELECT * FROM text_data;
```

3. **Create vector embeddings from the data table**  
   - Use `CREATE TABLE` with the `AI_TextEmbeddings` function to generate embeddings from your data, specifying the model, and store the results in the new table.

```
CREATE TABLE text_embeddings AS (
    SELECT * FROM AI_TextEmbeddings(
        ON text_data AS InputTable
        USING
        authorization(AWSEmbeddingsAuth)
        region('us-west-2')
        apitype('aws')
        modelname('amazon.titan-embed-text-v1')
        textcolumn('text')
        outputformat('vector')
    ) as DT
) WITH DATA;
```

By following these steps, we will successfully create vector table - `text_embeddings`.

```
-- View embeddings in table
SELECT * FROM text_embeddings;
```

### 🔍 Performing a Similarity Search in `text_embeddings`  

A **similarity search** enables you to find the most relevant entries in your vector table based on a given input. By creating a question table, and then embedding it, you can retrieve the closest matching results using your chosen vector search algorithm comparing with the reference table `text_embeddings`.

#### 🛠 How It Works  

1. **Create a table for the question**  
   - Use `CREATE TABLE` to define a new table called `query_table` with two columns:  
     - `id`: The primary key  
     - `query`: a text field to store the question itself  
   This table will hold the questions you want to search for in the vector store.

2. **Insert and embed the question in the table**  
   - Use `INSERT INTO` to add your question as a new row in the `query_table`.  
   - After inserting, generate the vector embedding for the question using the `AI_TextEmbeddings` function.

3. **Run a vector similarity search between the query embedding and the reference embeddings**  
   - Use `TD_VECTORDISTANCE` function to compare the embedded question with all reference embeddings. The result will return the **top_k** most similar entries, each with a similarity score, showing how closely each entry matches the input question.

4. **Use the id of the closest reference text to display the text**

This process enables efficient retrieval of information based on vector embeddings, making it useful for applications such as **semantic search, recommendation systems, and document retrieval**. 🚀  

```
CREATE TABLE query_table (
    id INTEGER GENERATED ALWAYS AS IDENTITY NOT NULL,
    query VARCHAR(1000),
    PRIMARY KEY (id)
);
```

```
INSERT INTO query_table (query) VALUES
('What is the meaning of life?');
```

```
CREATE TABLE query_embeddings AS (
    SELECT * FROM AI_TextEmbeddings(
        ON query_table AS InputTable
        USING
        authorization(AWSEmbeddingsAuth)
        region('us-west-2')
        apitype('aws')
        modelname('amazon.titan-embed-text-v1')
        textcolumn('query')
        outputformat('vector')
    ) as DT
) WITH DATA;
```

```
-- View the query embedding
SELECT * FROM query_embeddings;
```

```
SELECT target_id, reference_id, distancetype, cast(distance as decimal(36,15)) as distance 
FROM TD_VECTORDISTANCE (
    ON query_embeddings as TargetTable
    ON text_embeddings as ReferenceTable Dimension
    USING
        TargetIDColumn('id')
        TargetFeatureColumns('Embedding')
        RefIDColumn('id')
        RefFeatureColumns('Embedding')
        DistanceMeasure('euclidean')
        topk(5)
        EmbeddingSize(1536)
) as dt
ORDER BY distance ASC;
```

```
-- View the text that corresponds most closely with the question
    
SELECT text_data.text, text_data.id
FROM text_data
WHERE text_data.id = 3;
```

<b style="font-size:30px; font-family:Arial; color:#00233C;">3. Using the Existing Vector</b>

### 🔄 Using an Existing Vector Store Table  

If you have an existing vector store table in your database, you can reuse it directly in your SQL workflows. This allows you to perform similarity searches and other operations on the stored embeddings.

#### 🛠 Steps to Use an Existing Vector Store Table  

1. **Identify the Name of the Vector Store Table**  
   - Specify the name (e.g., `text_embeddings`) of the existing table that contains your vector embeddings.
   - This table should already have the necessary vector columns generated from previous embedding steps.

2. **Query or Search the Table**  
   - Use SQL queries or vector search functions to interact with the existing vector store table and perform similarity searches or other operations as needed.
   - In this section, we will demonstrate how to use the, `TD_VectorNormalize` function to normalize the embeddings, `TD_Kmeans` function for clustering, and the `TD_HNSW` function for approximate nearest neighbor search on your existing vector store table.

<div style="font-size:15px;font-family:Arial;color:#00233C;background:#fff8e6;padding:12px;border-radius:5px;border:1px solid #ffd700">
<b>Note:</b> The SQL only version does not include the update function. The python version, described in the first pre sales enablement training, includes the update and status capabilities.

#### 🧭 TD_Vectornormalize

`TD_VectorNormalize` is a Teradata SQL function that transforms input vectors into unit vectors, ensuring each has a Euclidean norm of 1. This normalization is essential for enabling accurate similarity comparisons, especially when using cosine distance in vector search workflows. We will use teh normalized embeddings in the Kmeans function.

```
CREATE TABLE text_embeddings_normalized AS (
    SELECT *
    FROM TD_Vectornormalize(
        ON text_embeddings AS InputTable
        USING
            IDColumns('id')
            TargetColumns('Embedding')
            Accumulate('text', 'Message')
            Approach('UNITVECTOR')
            EmbeddingSize(1536)
    ) AS DT) WITH DATA;
```

```
SELECT TOP 5 * FROM text_embeddings_normalized;
```

### 🔍 Performing Similarity Search on the Existing Vector Store Table

Once you've located the table, you can seamlessly perform a **similarity search**, and ask to retrieve the most relevant information.


#### 🎯 Kmeans algorithm

The K-means algorithm groups vectors into clusters by iteratively minimizing distances between points and cluster centroids. It enables efficient approximate similarity search by comparing vectors only to nearby cluster centers rather than all data points.

1. **Create the initial centroids tables**  
   - Execute the query to create the initial centroids table.

```
CREATE TABLE kmeans_initial_centroids_table as (
    select * 
    from text_embeddings_normalized
    where id in (2,4)
) with data;
```

```
SELECT * FROM kmeans_initial_centroids_table;
```

2. **Use the `TD_Kmeans` function**  

```
CREATE TABLE kmeans_centroids AS (
    SELECT * FROM TD_KMeans (
        ON text_embeddings_normalized AS InputTable
        ON (SELECT id, Embedding FROM kmeans_initial_centroids_table) AS InitialCentroidsTable DIMENSION
        USING
        IDColumn('id')
        TargetColumns('Embedding')
        StopThreshold(0.0395)
        MaxIterNum(3)
        -- OutputClusterAssignment('true')
    ) AS DT) WITH DATA;
```

3. **Select from the centroids table and view the Kmeans query**  

```
SELECT * FROM kmeans_centroids;
```

4. **Finally, use the `TD_KMEANSPREDICT` function** 
    - The output shows, for each input vector, which cluster it belongs to and how far it is from the centroid. 

```
SELECT * FROM TD_KMeansPredict (
ON text_embeddings_normalized AS InputTable
ON kmeans_centroids AS ModelTable DIMENSION
USING
OutputDistance('true')
-- Accumulate('id', 'text')
)AS dt
```

#### ⚡ HNSW Algorithm

The HNSW (Hierarchical Navigable Small World) algorithm organizes vectors into a graph structure to enable fast and scalable approximate nearest neighbor searches. Instead of comparing every vector, HNSW efficiently navigates the graph to quickly find the most similar vectors.

1. **Create the HNSW index table with `TD_HNSW` function**
   - Create index from your vector embeddings. 

```
SELECT * FROM TD_HNSW(
    ON text_embeddings_normalized AS InputTable
    OUT PERMANENT TABLE ModelTable(text_embeddings_index)
    USING
    IDColumn('id')
    VectorColumn('Embedding')
    Seed(1234)
    EfConstruction(32)
    NumConnPerNode(32)
    MaxNumConnPerNode(32)
    DistanceMeasure('euclidean')
    EmbeddingSize(1536)
    ApplyHeuristics('true')
) AS DT;
```

2. **View the created HNSW index**  

```
SELECT * FROM text_embeddings_index;
```

3. **Use `TD_HNSWPREDICT` to perform an approximate nearest neighbor search** 

```
SELECT * FROM TD_HNSWPredict (
 ON text_embeddings_index as ModelTable
 ON text_embeddings_normalized as InputTable DIMENSION
 USING
 IdColumn('id') 
 VectorColumn('Embedding')
) as alias;
```

4. **The `TD_HNSWSummary` function takes the ModelTable generated from TD_HNSW function as input and converts the ModelTable data into readable format.** 

```
SELECT * FROM TD_HNSWSummary (
    ON text_embeddings_index AS ModelTable
) AS ALIAS;
```

### 📂 Creating a Vector Store using PDF files.  

<div style="font-size:15px;font-family:Arial;color:#00233C;background:#fff8e6;padding:12px;border-radius:5px;border:1px solid #ffd700">
<b>Note:</b> Parsing a PDF and loading its content into a database table is covered in the Unstructured.io notebook. An example of this process is described in the Vector Store + Unstructured notebook included later in this pre sales enablement training.

For this notebook, we will use data that has already been extracted from a PDF and stored in a database, `TD_Vector`, in a table named `pdf_elements`.

1. **Create table of Vector Embeddings (`pdf_embeddings`) using data from a PDF**
   - Using the `AI_TextEmbeddings` function for the embeddings.

```
CREATE TABLE pdf_embeddings AS (
    SELECT * FROM AI_TextEmbeddings(
        ON TD_Vector.pdf_elements AS InputTable
        USING
        authorization(AWSEmbeddingsAuth)
        region('us-west-2')
        apitype('aws')
        modelname('amazon.titan-embed-text-v1')
        textcolumn('text')
        outputformat('vector')
    ) as DT
) WITH DATA;
```

2. **View the embeddings in the table from the PDF**

```
SELECT TOP 5 * FROM pdf_embeddings WHERE id >= 5;
```

#### ↔️ Vector Distance Semantic Similarity Search

Once you have the vector store table with the generated embeddings from the PDF, you can use the `TD_VECTORDISTANCE` function to perform a semantic similarity search. It compares the embedding of your query against all embeddings in the PDF table. 

#### 🛠 Steps to Retrieve Most Similar Results

1. **Prepare the Query**  
   - Create a table to store your query and insert the desired question.
   - Use the `AI_TextEmbeddings` function to generate an embedding for the query. Be sure to use the same model that was used to create the embeddings for the PDF data.

2. **Use `TD_VECTORDISTANCE` function**  
   - Use the **pdf_embeddings** as the Reference table and **query_table_pdf** as the Target table.
   -  The distance measure used is **euclidean**

```
CREATE TABLE query_table_pdf (
    id INTEGER GENERATED ALWAYS AS IDENTITY NOT NULL,
    query VARCHAR(1000),
    PRIMARY KEY (id)
);
```

```
INSERT INTO query_table_pdf (query) VALUES
('How can I optimize dense retrieval processing?');
```

```
CREATE TABLE query_embeddings_pdf AS (
    SELECT * FROM AI_TextEmbeddings(
        ON query_table_pdf AS InputTable
        USING
        authorization(AWSEmbeddingsAuth)
        region('us-west-2')
        apitype('aws')
        modelname('amazon.titan-embed-text-v1')
        textcolumn('query')
        outputformat('vector')
    ) as DT
) WITH DATA;
```

```
SELECT target_id, reference_id, distancetype, cast(distance as decimal(36,15)) as distance FROM TD_VECTORDISTANCE (
    ON query_embeddings_pdf as TargetTable
    ON pdf_embeddings as ReferenceTable Dimension
    USING
        TargetIDColumn('id')
        TargetFeatureColumns('Embedding')
        RefIDColumn('id')
        RefFeatureColumns('Embedding')
        DistanceMeasure('euclidean')
        topk(5)
        EmbeddingSize(1536)
) as dt
ORDER BY distance ASC;
```

3. **View the text that corresponds most closely with the question** 

```
SELECT pdf_embeddings.text, pdf_embeddings.id
FROM pdf_embeddings
WHERE pdf_embeddings.id = 102;
```
